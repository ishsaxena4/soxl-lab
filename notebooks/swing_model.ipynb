```python
index = 100
soxl = 100
soxs = 100
```

```python
daily_returns = [0.10, -0.0909]

for day, r in enumerate(daily_returns, start=1):
    index *= (1 + r)
    soxl *= (1 + 3*r)
    soxs *= (1 - 3*r)

    print (f"Day {day}")
    print (f" Index: {index:.2f}")
    print (f" SOXL: {soxl:.2f}")
    print  (f" SOXS: {soxs:.2f}")
    print()
```
